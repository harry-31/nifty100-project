```python
import pandas as pd

companies = pd.read_excel("../data/raw/companies.xlsx", header=1)

ids = [
    "AGTL",
    "ULTRACEMCO",
    "UNIONBANK",
    "UNITDSPR",
    "VBL",
    "VEDL",
    "WIPRO",
    "ZOMATO",
    "ZYDUSLIFE"
]

print("Total companies:", len(companies))

for cid in ids:
    print(f"\nSearching for: {cid}")
    result = companies[
        companies["id"].astype(str).str.contains(cid, case=False, na=False)
        | companies["company_name"].astype(str).str.contains(cid, case=False, na=False)
    ]
    print(result[["id", "company_name"]])
```
